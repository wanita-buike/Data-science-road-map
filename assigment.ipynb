# **Section 3: Pandas - Building and Exploring a Dataset**

### **Step 1 : Creating Student Dataset**

```python
import pandas as pd
```

```python
student_data = {
    "StudentID":    ["S010","S011","S012","S013","S014","S015","S016","S017"],
    "Name":         ["Wanita Bukie","John Selman","David Joe","Michael James","Vera Judge", "Jeffrey cole" ,"Brian Moon",
                     "Wisdom Karen" ],
    "Age":          [20, 22, 21, 23, 22, 24, 23, 24,],
    "Department":   ["Medicine","Dentistry","Radiography", "Pharmacy","Medical Lab","Law", "Nursing", "Accountancy"],
    "Score":        [ 70, 55, 78, 58, 62, 91, 64, 85,],
    "Grade":        ["A","C","A","C","B","A","B","A"]
}
```

```python
df = pd.DataFrame(student_data)
```

```python
df.shape # (rows, columns)
```

**Text Cell explaning the output Of Step 3**

The result from df.shape shows the number of rows and columns in the DataFrame.

The first number represents the number of rows (students) in the dataset,

The second number represents the number of columns ie; StudentID, Name, Age, Department, Score, and Grade.
