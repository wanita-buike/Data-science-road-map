## SECTION 3 ; Feature Scaling

### Question 9

### What is Feature Scaling and Why is it Necessary?

Feature scaling is the transforming of numerical data into the same range, usually between 0 and 1.

Feature scaling is necessary because many machine learning algorithms such as Linear Regression, and Gradient Descent-based models perform better and faster when data is scaled.

### What Problem Would Arise Without Scaling?

If the housing dataset is used without scaling, features with large values like total_rooms or population will dominate features with smaller values like median_income which wil cause the model to learn incorrectly and produce poor predictions.

### Question 10


```python
# Min-Max scaling on a single column
```

```python
cols_to_scale = ['housing_median_age', 'total_rooms', 'median_income', 'median_house_value']
```

```python
for col in cols_to_scale:
    col_min = df[col].min()
    col_max = df[col].max()
    df[col + '_scaled'] = (df[col] - col_min) / (col_max - col_min)
```

### Question 11

```python
scaled_cols = [
    'housing_median_age_scaled',
    'total_rooms_scaled',
    'median_income_scaled',
    'median_house_value_scaled'
]

for col in scaled_cols:
    print(f"{col}: min = {df[col].min():.1f}, max = {df[col].max():.1f}")
```

Looking at median_income and median_income_scaled side by side, the relative
ordering of the values is completely preserved — a row with a higher income still
has a higher scaled value. The only thing that changed is the range. The original

a scaled value of exactly 0.0, and the row with the highest income gets exactly
